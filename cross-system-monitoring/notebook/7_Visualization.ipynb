```python
from pyspark.sql import SparkSession
from delta import configure_spark_with_delta_pip

builder = (
    SparkSession.builder
    .appName("DataTrustPlatform")
    .master("local[*]")
    .config(
        "spark.jars.packages",
        "io.delta:delta-spark_2.12:3.2.0"
    )
    .config(
        "spark.sql.extensions",
        "io.delta.sql.DeltaSparkSessionExtension"
    )
    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog"
    )
)

spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
from pyspark.sql import SparkSession
import matplotlib.pyplot as plt
import pandas as pd

spark = SparkSession.builder.getOrCreate()
```

```python
dashboard = spark.read.format("delta").load("../gold/dashboard_data")
```

```python
dashboard_pd = dashboard.toPandas()

dashboard_pd
```

Dashboard Summary

```python
plt.figure(figsize=(10,5))

plt.bar(
    dashboard_pd["Metric"],
    dashboard_pd["Value"]
)

plt.xticks(rotation=45)

plt.title("Cross-System Monitoring Dashboard")

plt.xlabel("Metrics")

plt.ylabel("Value")

plt.tight_layout()

plt.show()
```

System Record Comparison

```python
records = dashboard_pd.iloc[:3]

plt.figure(figsize=(7,5))

plt.bar(
    records["Metric"],
    records["Value"]
)

plt.title("Records Across Systems")

plt.ylabel("Count")

plt.show()
```

Data Quality Issues

```python
quality = dashboard_pd.iloc[3:6]

plt.figure(figsize=(7,5))

plt.bar(
    quality["Metric"],
    quality["Value"]
)

plt.title("Data Quality Issues")

plt.ylabel("Count")

plt.show()
```

Pie Chart

```python
plt.figure(figsize=(6,6))

plt.pie(
    quality["Value"],
    labels=quality["Metric"],
    autopct="%1.1f%%"
)

plt.title("Distribution of Data Quality Issues")

plt.show()
```

Trust Score

```python
trust = dashboard_pd[
    dashboard_pd["Metric"]=="Trust Score"
]

plt.figure(figsize=(4,5))

plt.bar(
    trust["Metric"],
    trust["Value"]
)

plt.ylim(0,100)

plt.title("Overall Trust Score")

plt.show()
```

Horizontal Bar Chart

```python
plt.figure(figsize=(9,5))

plt.barh(
    dashboard_pd["Metric"],
    dashboard_pd["Value"]
)

plt.title("Dashboard Metrics")

plt.show()
```

KPI Summary

```python
crm = dashboard_pd.loc[
    dashboard_pd["Metric"]=="CRM Records",
    "Value"
].values[0]

billing = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Billing Records",
    "Value"
].values[0]

analytics = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Analytics Records",
    "Value"
].values[0]

missing = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Missing Records",
    "Value"
].values[0]

duplicates = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Duplicate Records",
    "Value"
].values[0]

drift = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Drift Records",
    "Value"
].values[0]

trust = dashboard_pd.loc[
    dashboard_pd["Metric"]=="Trust Score",
    "Value"
].values[0]

print("="*50)

print("Cross-System Monitoring Dashboard")

print("="*50)

print("CRM Records        :", crm)

print("Billing Records    :", billing)

print("Analytics Records  :", analytics)

print("Missing Records    :", missing)

print("Duplicate Records  :", duplicates)

print("Drift Records      :", drift)

print("Trust Score        :", trust)

print("="*50)
```
