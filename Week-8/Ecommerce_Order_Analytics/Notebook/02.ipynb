```python
import sqlite3
import pandas as pd
import os

print("Libraries Imported Successfully")
```

```python
customers = pd.read_csv("data/cleaned/customers_cleaned.csv")
products = pd.read_csv("data/cleaned/products_cleaned.csv")
orders = pd.read_csv("data/cleaned/orders_cleaned.csv")
order_items = pd.read_csv("data/cleaned/order_items_cleaned.csv")

print("Cleaned Data Loaded Successfully")
```

```python
os.makedirs("database", exist_ok=True)

conn = sqlite3.connect("database/ecommerce.db")

cursor = conn.cursor()

print("Database Created Successfully")
```

```python
cursor.execute("""

CREATE TABLE IF NOT EXISTS customers(

customer_id TEXT PRIMARY KEY,

customer_name TEXT,

email TEXT,

registration_date DATE,

customer_type TEXT

)

""")

conn.commit()

print("Customers Table Created")
```

```python
cursor.execute("""

CREATE TABLE IF NOT EXISTS products(

product_id TEXT PRIMARY KEY,

product_name TEXT,

category TEXT,

subcategory TEXT,

cost_price REAL

)

""")

conn.commit()

print("Products Table Created")
```

```python
cursor.execute("""

CREATE TABLE IF NOT EXISTS orders(

order_id TEXT PRIMARY KEY,

customer_id TEXT,

order_date TEXT,

status TEXT,

region_code TEXT

)

""")

conn.commit()

print("Orders Table Created")
```

```python
cursor.execute("""

CREATE TABLE IF NOT EXISTS order_items(

item_id INTEGER PRIMARY KEY,

order_id TEXT,

product_id TEXT,

quantity INTEGER,

unit_price REAL,

discount_percent REAL

)

""")

conn.commit()

print("Order Items Table Created")
```

```python
customers.to_sql("customers", conn,
if_exists="replace",
index=False)

print("Customers Loaded")
```

```python
products.to_sql(

"products",

conn,

if_exists="replace",

index=False

)

print("Products Loaded")
```

```python
orders.to_sql(

"orders",

conn,

if_exists="replace",

index=False

)

print("Orders Loaded")
```

```python
order_items.to_sql(

"order_items",

conn,

if_exists="replace",

index=False

)

print("Order Items Loaded")
```
