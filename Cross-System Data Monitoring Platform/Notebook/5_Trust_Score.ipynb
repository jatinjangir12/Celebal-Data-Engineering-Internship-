reading data

```python
missing = spark.read.format("delta").load("/Workspace/Cross-System Data Monitoring Platform/gold/missing_records")

duplicates = spark.read.format("delta").load("/Workspace/Cross-System Data Monitoring Platform/gold/duplicates")

drift = spark.read.format("delta").load("/Workspace/Cross-System Data Monitoring Platform/gold/drift_report")
```

count

```python
missing_count = missing.count()

duplicate_count = duplicates.count()

drift_count = drift.filter(drift.difference > 100).count()
```

load drift

```python
drift = spark.read.format("delta").load("/Workspace/Cross-System Data Monitoring Platform/gold/drift_report")
```

dashboard score

```python

score = 100.0

score -= (missing_count * 0.005)

score -= (duplicate_count * 0.01)

score -= (drift_count * 0.05)

score = max(round(score, 1), 0.0)

print(f"Dashboard Trust Score = {score}/100")
```

saving trust score

```python
trust = spark.createDataFrame([(score,)], ["trust_score"])

trust.write.format("delta").mode("overwrite").save("/Workspace/Cross-System Data Monitoring Platform/gold/trust_score")
```
