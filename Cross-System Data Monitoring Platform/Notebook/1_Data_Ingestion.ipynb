reading the datasets

```python
crm = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Workspace/Cross-System Data Monitoring Platform/data/crm_dataset.csv")
)

billing = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Workspace/Cross-System Data Monitoring Platform/data/billing_dataset.csv")
)

analytics = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Workspace/Cross-System Data Monitoring Platform/data/analytics_dataset.csv")
)
```

schema

```python
print("CRM Schema")
crm.printSchema()

print("Billing Schema")
billing.printSchema()

print("Analytics Schema")
analytics.printSchema()
```

```python
print("staring rows")
crm.show(5)

billing.show(5)

analytics.show(5)
```

saving filter data into delta table

```python
crm.write.format("delta").mode("overwrite").save("/Workspace/Cross-System Data Monitoring Platform/bronze/crm")

billing.write.format("delta").mode("overwrite").save("/Workspace/Cross-System Data Monitoring Platform/bronze/billing")

analytics.write.format("delta").mode("overwrite").save("/Workspace/Cross-System Data Monitoring Platform/bronze/analytics")
```
